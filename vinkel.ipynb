```python
def solar_panel_flux(daily_data, theta_p, phi_p):
    shape = daily_data['azimuth'].to_numpy().shape
    theta_panel, phi_panel = np.full(shape, theta_p), np.full(shape, phi_p) # Solar panel static
    flux = solar_panel_projection(
        daily_data['zenith'].to_numpy(), 
        daily_data['azimuth'].to_numpy(), 
        theta_panel, 
        phi_panel
    )

    return flux
```

```python
A = 2.8 # m^2
A_0 = 0.5
S_0 = 1100 # W/m^2
G_avg = S_0 * A_0 # k/m^2

p_rated = 605 / (1000 * A)

def solar_panel_power(daily_data, theta_p, phi_p, delta_time="h"): 
    flux = solar_panel_flux(daily_data, theta_p, phi_p)
    scale = 60 * (1 if delta_time == "Min" else 60)
    integral_value = integrate.simpson(flux, dx=scale) * p_rated * G_avg
    return integral_value # joules per m^2

def J_to_kWh(J):
    return J / (3.6 * 10**6)
```

```python
chosen_date = "2024-04-20"
daily_data = get_solar_position(chosen_date, chosen_date)

p_april = solar_panel_power(daily_data, 45, 180) # Joules per m^2
print(f'{p_april} J/m^2')
print(f'{p_april * A} J per panel')
p_kW_per_m_sqrt = J_to_kWh(p_april) # kW / m^2
print(f'{p_kW_per_m_sqrt} kWh / m^2')
print(f'{p_kW_per_m_sqrt * A} kWh per panel')
```
