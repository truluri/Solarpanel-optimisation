```python
def inner(u_s: Matrix,u_p: Matrix):
    '''
    Computes the inner product of two vectors of same length.
    '''
    
    return u_s.dot(u_p, conjugate_convention = 'right')

MutableDenseMatrix.inner = inner
ImmutableDenseMatrix.inner = inner
```
